## 2. Understanding the Dataset

The LA Crime dataset contains a large collection of reported incidents from across Los Angeles. Each row represents a single crime report. The dataset includes a mix of time information, location details, crime categories, and basic victim attributes. Before exploring any patterns, it is important to understand what the dataset provides and how it is structured.

Key fields in the dataset include:

- **`DATE OCC`** which records the date of the incident  
- **`TIME OCC`** which records the time of day  
- **`AREA NAME`** which identifies the reporting area  
- **`Crm Cd Desc`** which describes the type of crime  
- **`Vict Age`** which records the age of the victim  
- **`Vict Sex`** which records the victim's gender  

Some fields required cleaning or transformation before they could be used. For example:

- `TIME OCC` was converted into an `HOUR` column  
- unused columns were removed  
- categorical fields were prepared for analysis  
- a new target variable called `is_violent` was created  

The goal of this section is to give a clear picture of what information is available and how it supports the analysis that follows.
